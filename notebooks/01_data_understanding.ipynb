# Phase 1: Data Foundation — Step 1.1: Data Understanding

This notebook performs the initial data understanding for the 5 active raw datasets of the Agentic HRMS project, reporting their shape, data types, missing value counts, duplicate counts, any ID-related columns, and the class balance for employee attrition.

```python
import os
import pandas as pd

raw_dir = os.path.join("data", "raw")
print(f"Checking files in: {os.path.abspath(raw_dir)}")
```

## 1. Employee Attrition Dataset (`employee_attrition.csv`)
* **Role**: Attrition ML source + Employee Master
* **Description**: Feeds the attrition model and stands in as the core employee table.

```python
file_path = os.path.join(raw_dir, "employee_attrition.csv")
df_attrition = pd.read_csv(file_path)

print(f"Shape: {df_attrition.shape}")
print("\nData Types:")
print(df_attrition.dtypes)

print("\nMissing Values:")
missing = df_attrition.isnull().sum()
print(missing[missing > 0] if missing.sum() > 0 else "No missing values")

print(f"\nDuplicate Count: {df_attrition.duplicated().sum()}")

id_cols = [c for c in df_attrition.columns if "id" in c.lower()]
print(f"Columns containing 'id': {id_cols}")

print("\nAttrition Class Balance:")
print(df_attrition["Attrition"].value_counts())
print(df_attrition["Attrition"].value_counts(normalize=True))
```
