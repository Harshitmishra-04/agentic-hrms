# Phase 2: ML Attrition Model — Step 2.5: Model Versioning

This notebook saves our finalized balanced Logistic Regression pipeline under the versioned directory `models/v1/` and creates a corresponding metadata registry JSON file (`metadata.json`). This ensures model lineage, transparency, and traceability for downstream integrations.

```python
import os
import shutil
import json
from datetime import datetime

v1_dir = os.path.join("models", "v1")
os.makedirs(v1_dir, exist_ok=True)
```

## 1. Copy Finalized Pipeline to v1 Registry

```python
src_pipeline = os.path.join("models", "attrition_pipeline.joblib")
dest_pipeline = os.path.join(v1_dir, "attrition_pipeline.joblib")

shutil.copyfile(src_pipeline, dest_pipeline)
print(f"Model copied successfully to {dest_pipeline}")
```

## 2. Write model metadata.json
We log model properties, algorithms, training details, and test evaluation metrics to document model versioning.

```python
metadata = {
    "model_name": "attrition_predictor",
    "version": "1.0.0",
    "algorithm": "LogisticRegression (Balanced Weights)",
    "training_date": datetime.now().strftime("%Y-%m-%d"),
    "roc_auc": 0.8007,
    "f1_score": 0.5067,
    "precision": 0.3689,
    "recall": 0.8085,
    "decision_threshold": 0.40,
    "decision_threshold_note": "Must be applied explicitly via predict_proba() >= threshold in any service code, since the sklearn pipeline's own .predict() defaults to 0.5.",
    "selection_rationale": "Balanced Logistic Regression with a decision threshold of 0.40 was chosen over the Regularized Random Forest alternative (Test Recall 68.09%, Precision 45.07%, F1 54.24%, ROC-AUC 0.7748) because maximizing recall was prioritized to catch more at-risk employees (80.85% vs 68.09%). Additionally, Logistic Regression maintains a higher ROC-AUC (0.8007 vs 0.7748). Regularized Random Forest remains documented as a viable v2 alternative if precision becomes the priority later.",
    "known_limitation": "~2 of 3 flagged employees are false positives; downstream HR workflow using this model should treat flags as a prompt for review, not an automatic action."
}

metadata_path = os.path.join(v1_dir, "metadata.json")
with open(metadata_path, "w") as f:
    json.dump(metadata, f, indent=4)

print(f"Model metadata saved successfully to {metadata_path}:")
print(json.dumps(metadata, indent=4))
```
