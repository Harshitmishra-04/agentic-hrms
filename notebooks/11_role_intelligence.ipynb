# Phase 3: Workforce Intelligence — Step 3.2: Role Intelligence Reference

This notebook processes the occupation master dataset (`data/processed/occupation_master.csv`) into a standardized role reference table. We assign a clean, unique alphanumeric identifier (`role_id`) to each O*NET-SOC code, rename and clean columns, and export the resulting table to `data/processed/role_reference.csv` for downstream lookup.

```python
import os
import pandas as pd

proc_dir = os.path.join("data", "processed")
print(f"Processed directory: {os.path.abspath(proc_dir)}")
```

## 1. Load Occupation Master

```python
df_occ = pd.read_csv(os.path.join(proc_dir, "occupation_master.csv"))
print(f"Occupation Master Shape: {df_occ.shape}")
print(df_occ.head())
```

## 2. Generate Clean role_id and Rename Columns
We generate unique IDs formatted as `ROLE_XXXX` (zero-padded) based on index and standardize the column names.

```python
df_occ = df_occ.reset_index(drop=True)
df_occ["role_id"] = [f"ROLE_{i+1:04d}" for i in range(len(df_occ))]

df_role_ref = df_occ[[
    "role_id",
    "O*NET-SOC Code",
    "Title",
    "Description"
]].rename(columns={
    "O*NET-SOC Code": "onet_soc_code",
    "Title": "role_title",
    "Description": "role_description"
})

print("Sample role reference entries:")
print(df_role_ref.head())
```

## 3. Verify Integrity

```python
assert df_role_ref["role_id"].is_unique, "role_id is not unique!"
assert df_role_ref["onet_soc_code"].is_unique, "onet_soc_code is not unique!"
print("Integrity checks passed: all IDs are unique.")
```

## 4. Save to Processed Data

```python
output_path = os.path.join(proc_dir, "role_reference.csv")
df_role_ref.to_csv(output_path, index=False)
print(f"Saved role reference table successfully to {output_path}")
```
